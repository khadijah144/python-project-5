```python
%%markdown
# Contact Book
```

```python
contacts = {}
```

```python
%%markdown
## CONTACT MENU
#### 1 = Add contact
#### 2 = Search Contact
#### 3 = Exit
```

```python
while True:
    print("")
    print("Enter 1 to Add contacts")
    print("Enter 2 to Search Contact")
    print("Enter 3 to Exit")

    choice = input("Enter your choice: ").strip()
    if choice == "1":
        name = input("Enter contact name: ").strip().title()
        phone = input("Enter phone number: ").strip()
        email = input("Enter email address: ").strip()
        if name == "1" or phone == "" or email == "":
            print("All information is required.")
        else:
            contacts[name] = {
                "phone" : phone,
                "email" : email
            }

            print("Contact added successfully.")
    elif choice == "2":
        searchname = input("Enter contact name to search: ").strip().title()
        if searchname in contacts:
            print("Contact Found")
            print("Name:", searchname)
            print("Phone number:", contacts[searchname]["phone"])
            print("Email:", contacts[searchname]["email"])
        else:
            print("Contact not found.")
    elif choice == "3":
        print("")
        print("Contact book closed")
        break

    else:
        print("")
        print("Invalid choice. Enter 1, 2 or 3.")
```
